```python
#Check Fraud Distribution
```

```python
print(df["Fraud"].value_counts())
```

```python
#Total Transaction Amount
```

```python
print(df["Amount"].sum())
```

```python
#Fraud Transaction Amount
```

```python
fraud_amount = df[df["Fraud"] == 1]["Amount"].sum()

print(fraud_amount)
```

```python
#Genuine Transaction Amount
```

```python
genuine_amount = df[df["Fraud"] == 0]["Amount"].sum()

print(genuine_amount)
```

```python
#Highest Transaction
```

```python
print(df["Amount"].max())
```

```python
#Location-wise Transactions
```

```python
print(df.groupby("Location")["Amount"].sum())
```

```python
#Fraud Transactions Only
```

```python
fraud_data = df[df["Fraud"] == 1]

print(fraud_data)
```

```python
#Fraud Percentage
```

```python
fraud_percent = (
    len(df[df["Fraud"] == 1]) /
    len(df)
) * 100

print(fraud_percent)
```
