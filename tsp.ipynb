```python
# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 2. Create Dataset
data = {
    'Age': [22, 38, 26, 35, 28, 2, 54, 27, 14, 40],
    'Sex': [0, 1, 1, 1, 0, 0, 0, 1, 1, 0],
    'Fare': [7, 71, 8, 53, 13, 21, 51, 11, 30, 27],
    'Survived': [0, 1, 1, 1, 0, 1, 0, 1, 1, 0]
}

df = pd.DataFrame(data)
```

```python
# 3. Display Dataset
print("Titanic Dataset:")
print(df)
```

```python
# 4. Check Null Values
print("\nNull Values:")
print(df.isnull().sum())
```

```python
# 5. Dataset Information
print("\nDataset Information:")
print(df.info())
```

```python
# 6. Bar Chart
df['Survived'].value_counts().plot(kind='bar')
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
# 7. Pie Chart
df['Survived'].value_counts().plot(
    kind='pie', autopct='%1.1f%%'
)
plt.title("Titanic Survival Percentage")
plt.ylabel("")
plt.show()
```

```python
# 8. Scatter Plot
plt.scatter(df['Age'], df['Fare'],
            c=df['Survived'])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
# 9. Heatmap
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
